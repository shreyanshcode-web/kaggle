```python
from nltk.tokenize import WhitespaceTokenizer

def whitespace_tokenize(text):
    return WhitespaceTokenizer().tokenize(text)
text = "Hello, my name is shreyansh singh / i am a computer developere ."
tokens = whitespace_tokenize(text)
print(tokens)

name = "so what is the current market price of bitcoin ?"
tokens = whitespace_tokenize(name)
print(tokens)

```

```python
from nltk.tokenize import word_tokenize 

text = "hmm what is the new state of the matter ehich is evolveing here "
words = word_tokenize(text)
print(words)

text = "the weekneds plans the next tour to the  las vagas "
words = word_tokenize(text)
print(words)
```
